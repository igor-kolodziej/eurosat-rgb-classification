Thin report notebook for the saved EuroSAT RGB EDA artifacts.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, Markdown, display

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'outputs').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

EDA_METRICS = PROJECT_ROOT / 'outputs' / 'metrics' / 'eda'
EDA_FIGURES = PROJECT_ROOT / 'outputs' / 'figures' / 'eda'
```

```python
display(Markdown((EDA_METRICS / 'eda_summary.md').read_text()))
display(pd.read_csv(EDA_METRICS / 'class_counts.csv'))
display(json.loads((EDA_METRICS / 'image_properties.json').read_text()))
```

```python
for filename in [
    'class_counts.png',
    'representative_samples.png',
    'brightness_histogram.png',
    'confusable_classes.png',
]:
    display(Markdown(f'## {filename}'))
    display(Image(filename=EDA_FIGURES / filename))
```
