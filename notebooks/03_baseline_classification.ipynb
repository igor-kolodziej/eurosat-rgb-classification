# 03 Baseline Classification

Thin report notebook for the saved baseline training artifacts.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, Markdown, display

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'outputs').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

BASELINE_METRICS = PROJECT_ROOT / 'outputs' / 'metrics' / 'baseline'
BASELINE_FIGURES = PROJECT_ROOT / 'outputs' / 'figures' / 'baseline'
baseline = json.loads((BASELINE_METRICS / 'baseline_cnn_v0.json').read_text())
```

```python
display(Markdown((BASELINE_METRICS / 'baseline_interpretation.md').read_text()))
summary = pd.DataFrame([
    {
        'split': 'validation',
        'accuracy': baseline['validation']['accuracy'],
        'macro_f1': baseline['validation']['macro_f1'],
    },
    {
        'split': 'test',
        'accuracy': baseline['test']['accuracy'],
        'macro_f1': baseline['test']['macro_f1'],
    },
])
display(summary)
display(pd.DataFrame(baseline['history']))
```

```python
for filename in [
    'training_history_v0.png',
    'confusion_matrix_v0.png',
]:
    display(Markdown(f'## {filename}'))
    display(Image(filename=BASELINE_FIGURES / filename))
```
